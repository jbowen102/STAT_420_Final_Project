### Cross-validation and RMSE

#### base/full additive model

```r
mod_add_pwo_form = det_freq ~ (mean_num_observers + mean_start_time_hr + med_duration + mean_dist_km +
                                               dist_to_coast_mi + elevation_ft + dominant_eco_group +
                                               tmean_2024_C + ppt_2024_mm)
mod_add_pwo_train = lm(mod_add_pwo_form, data = ebird_features_df_pwo[ebird_features_df_pwo$type == "train", ], weight = n_checklists)
summary(mod_add_pwo_train)
```

```r
# RMSE Of model - trained w/ whole dataset! (artificially low)
sqrt(mean(resid(mod_add_pwo)^2))
```

```r
# RMSE of one manually-created fold
yhat = predict(mod_add_pwo_train, newdata = ebird_features_df_pwo[ebird_features_df_pwo$type == "test",])
y = ebird_features_df_pwo[ebird_features_df_pwo$type == "test",]$det_freq
e = y - yhat
sqrt(mean(e^2))
```

```r
# avg RMSE across all 10 folds
train_control_k10 <- caret::trainControl(method = "cv", number = 10)
mod_add_pwo_cv_fit <- caret::train(
  form = mod_add_pwo_form,
  data = ebird_features_df_pwo,
  weights = n_checklists,
  method = "lm",
  trControl = train_control_k10
)
mod_add_pwo_cv_fit
```

```r
mod_add_pwo_cv_fit$results[1,"RMSE"]
```

#### log-response

```r
mod_add_pwo_logy_form = log(det_freq + 0.1) ~ (mean_num_observers + mean_start_time_hr + med_duration + mean_dist_km +
                                               dist_to_coast_mi + elevation_ft + dominant_eco_group +
                                               tmean_2024_C + ppt_2024_mm)
mod_add_pwo_logy_train = lm(mod_add_pwo_logy_form, data = ebird_features_df_pwo[ebird_features_df_pwo$type == "train", ], weight = n_checklists)
summary(mod_add_pwo_logy_train)
```

```r
# RMSE Of model - trained w/ whole dataset! (artificially low)
transf_fitted = exp(predict(mod_add_pwo_logy)) - 0.1
residuals = transf_fitted - ebird_features_df_pwo$det_freq
sqrt(mean(residuals^2))
```

```r
# RMSE of one manually-created fold
yhat = predict(mod_add_pwo_logy_train, newdata = ebird_features_df_pwo[ebird_features_df_pwo$type == "test",])
y = ebird_features_df_pwo[ebird_features_df_pwo$type == "test",]$det_freq
e = y - (e^yhat-0.1)
sqrt(mean(e^2))
```

```r
# avg RMSE across all 10 folds
train_control_k10 <- caret::trainControl(method = "cv", number = 10, savePredictions = "final")
mod_add_pwo_logy_cv_fit <- caret::train(
  form = mod_add_pwo_logy_form,
  data = ebird_features_df_pwo,
  weights = n_checklists,
  method = "lm",
  trControl = train_control_k10
)

# RMSE on original detection-frequency scale
preds <- mod_add_pwo_logy_cv_fit$pred
preds$pred_original <- exp(preds$pred)-0.1
preds$obs_original  <- exp(preds$obs)-0.1

rmse_original <- sqrt(mean((preds$obs_original - preds$pred_original)^2))
rmse_original
# ChatGPT helped with this.
```
