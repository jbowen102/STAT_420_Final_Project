```r
mod_add_osp = lm(det_freq ~ mean_num_observers + mean_start_time_hr + med_duration + mean_dist_km +
                        dist_to_coast_mi + elevation_ft + dominant_eco_group +
                        tmean_2024_C + ppt_2024_mm,
                data = ebird_features_df_osp, weight = n_checklists)

summary(mod_add_osp)
```

```r
mod_int_osp = lm(det_freq ~ (mean_num_observers + mean_start_time_hr + med_duration + mean_dist_km +
                         dist_to_coast_mi + elevation_ft + dominant_eco_group +
                         tmean_2024_C + ppt_2024_mm)^2,
                data = ebird_features_df_osp, weight = n_checklists)
summary(mod_int_osp)
```

```r
mod_add_pwo = lm(det_freq ~ mean_num_observers + mean_start_time_hr + med_duration + mean_dist_km +
                        dist_to_coast_mi + elevation_ft + dominant_eco_group +
                        tmean_2024_C + ppt_2024_mm,
                data = ebird_features_df_pwo, weight = n_checklists)

summary(mod_add_pwo)
```

```r
mod_int_pwo = lm(det_freq ~ (mean_num_observers + mean_start_time_hr + med_duration + mean_dist_km +
                         dist_to_coast_mi + elevation_ft + dominant_eco_group +
                         tmean_2024_C + ppt_2024_mm)^2,
                data = ebird_features_df_pwo, weight = n_checklists)
summary(mod_int_pwo)
```

```r
mod_int_pwo = lm(det_freq ~ (mean_num_observers + mean_start_time_hr + med_duration + mean_dist_km +
                         dist_to_coast_mi + elevation_ft + dominant_eco_group +
                         tmean_2024_C + ppt_2024_mm)^2,
                data = ebird_features_df_pwo, weight = n_checklists)
summary(mod_int_pwo)
```
